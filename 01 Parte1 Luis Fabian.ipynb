## 4. Limpieza
Dejamos nombres simples, quitamos datos repetidos o vacíos y separamos a Colombia.

```python
# Nombres simples para las columnas
datos.columns = ["pais", "codigo", "año", "porcentaje"]

print("Datos vacíos encontrados:")
print(f"  - En el porcentaje: {datos['porcentaje'].isnull().sum()}")
print(f"  - En código de país: {datos['codigo'].isnull().sum()} (es normal: son regiones, no países)")
print(f"Filas repetidas: {datos.duplicated().sum()}")
```

```python
# Quitamos repetidos y vacíos
datos = datos.drop_duplicates()
datos = datos.dropna(subset=["porcentaje"])

# Nos quedamos con Colombia
colombia = datos[datos["pais"] == "Colombia"].copy()
colombia = colombia.sort_values("año")

print(f"✅ Limpieza lista. Quedaron {len(datos)} filas en total.")
if len(colombia) == 0:
    print("⚠️ No se encontraron datos de Colombia. Revisa el nombre del país en la tabla.")
else:
    print(f"🇨🇴 Colombia tiene {len(colombia)} años de datos, de {colombia['año'].min()} a {colombia['año'].max()}.")
colombia.head()
```
